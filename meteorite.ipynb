- The Scenario: You act as junior planetary scientists analyzing historical data of meteorite impacts and discoveries on Earth.
- Data Structures: You separate out specific columns, like the mass_grams, to see how it behaves as an isolated 1D Series compared to the broader 2D DataFrame context.
- Data I/O: You start by loading the raw data using pd.read_csv(). Later, you will export your filtered findings using .to_excel() or .to_json() for a museum curation team to use for an upcoming exhibit.
- Initial Data Inspection: Before doing anything, you must run .info() and .describe() to understand the sheer volume of data and check for missing values. You will use .head() to see the layout of the columns, and use .value_counts() to see which meteorite classification types (e.g., Chondrite, Iron) are the most frequently found.

```python
import pandas as pd
df = pd.read_csv('meteorites.csv')
print(df)
# Initail Data Inspection
df.info()
print(f"Describe : {df.describe()}")
print(f"Head : {df.head()}")
print(df['classification'].value_counts())

# Data Structure
mass = df['mass_grams']
print(f"Mass in grams : {mass}")
```

- Indexing and Selection: 
- Column/Label/Integer: Use dictionary-style selection to grab the fall_status column (whether it was seen falling or just found later). Use .iloc[] to extract the top 5 heaviest meteorites based on their sorted row position.
- Filtering: Apply multiple boolean conditions using & and | to find meteorites that are both greater than 50,000 grams in mass and were discovered in a specific region (e.g., "Antarctica" or "Sahara").

```python
#Column/Label/Integer
fall_status = df['fall_status'].head()
print(f"Fall status : {fall_status}")
value_top5_heaviest_meteorites = df.sort_values('mass_grams',ascending = False)
top5_heaviest_meteorites = value_top5_heaviest_meteorites.iloc[0:5]
print(f"Top 5 heaviest meteorites : {top5_heaviest_meteorites}")

#Filtering
bool_graterthan50000_and_specificregion = (mass > 50000) & (df['region'] == "sahara")
filtered_graterthan50000_and_specificregion = df[bool_graterthan50000_and_specificregion]
print(f"The mass greater than 50000 and discovered in sahara : {filtered_graterthan50000_and_specificregion}")
bool_graterthan50000_or_specificregion = (mass > 50000) | (df['region'] == "sahara")
filtered_graterthan50000_or_specificregion = df[bool_graterthan50000_or_specificregion]
print(f"The mass greater than 50000 or discovered in sahara : {filtered_graterthan50000_or_specificregion}")
df.to_excel("filtered_graterthan50000_or_specificregion.xlsx", index = False)
df.to_json("filtered_graterthan50000_or_specificregion.json", orient = "records")
```
